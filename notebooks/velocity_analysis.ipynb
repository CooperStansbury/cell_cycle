```python
%%time
scv.tl.rank_velocity_genes(
    adata,
    groupby='cluster_phase'
)

df = pd.DataFrame(adata.uns['rank_velocity_genes']['names'])
df.head()
```

```python
n_genes = 5

for column in df.columns:

    plt.rcParams['figure.dpi'] = 200
    plt.rcParams['figure.figsize'] = 4, 3.5
    
    scv.pl.scatter(
        adata, 
        df[column][:n_genes], 
        ylabel=column,
        color='cluster_phase',
        add_outline=True,
        frameon=False,
        size=10,
        linewidth=3,
        dpi=200,
    )
```
